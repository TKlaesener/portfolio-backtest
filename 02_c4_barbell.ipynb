```python
import openpyxl
import pandas as pd
import numpy as np

wb = openpyxl.load_workbook('data/raw/backtest_carteiras.xlsx', data_only=True)
ws = wb['C4 · BARBELL (NASSIM TALEB)']
```

```python
ativos = {
    "ouro": 3,          # coluna C
    "sp500": 7,          # coluna G
    "small_caps": 11,    # coluna K
    "emergentes": 15,    # coluna O
    "bil": 19,           # coluna S
}

linhas = []

for linha_excel in range(31, 276):  # C4 começa na linha 31, não 21 como o C1
    mes = ws.cell(row=linha_excel, column=2).value

    if mes is None:
        break

    registro = {"mes": mes}

    for nome_ativo, coluna_inicio in ativos.items():
        registro[f"{nome_ativo}_retorno"] = ws.cell(row=linha_excel, column=coluna_inicio).value
        registro[f"{nome_ativo}_peso"] = ws.cell(row=linha_excel, column=coluna_inicio + 1).value
        registro[f"{nome_ativo}_saldo"] = ws.cell(row=linha_excel, column=coluna_inicio + 2).value
        registro[f"{nome_ativo}_mov"] = ws.cell(row=linha_excel, column=coluna_inicio + 3).value

    registro["saldo_total"] = ws.cell(row=linha_excel, column=23).value   # coluna W
    registro["retorno_total"] = ws.cell(row=linha_excel, column=24).value  # coluna X

    linhas.append(registro)

print(f"Extraídos {len(linhas)} meses")
print(linhas[0])
print(linhas[-1])
```

```python
df = pd.DataFrame(linhas)
df["data"] = pd.to_datetime(df["mes"], format="%b/%Y")
df = df.drop(columns=["mes"]).set_index("data").sort_index()

df.to_csv("data/processed/c4_barbell.csv")
df.head()
```
